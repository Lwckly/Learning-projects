```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom
import math
from scipy.integrate import quad
```

```python
#grafico  de Pn(y) para uma probilidade de ocorrência de A sendo p e n repetições do processo
```

```python
p=0.5
n=100

# para a binomial a média de y é np e a variância é np(1-p)
# criar lista de valores que representam os diferentes numeros de ocorrencias de um evento
x = np.linspace(15, 100, 86)

# imprimindo a lista
print(x)

# gerando lista das probabilidades
y = binom.pmf(x, n, p)

print(y)
```

```python
# fazendo gráfico de Pn(y) x y
plt.xlabel('y',fontsize=18) # legenda eixo x
plt.ylabel('$P_n(y)$',fontsize=18) # legenda eixo y
plt.title('Probabilidade binomial para n=100 e p=0.5', fontsize=18)

# graficando a função
plt.bar(x,y,width=1)
```

```python
# parametros da gaussiana
mu=0
sigma=1
size=1000

# geração das variáveis da gaussiana
y=np.random.normal(mu,sigma,size)
```

```python
lista = []
plt.xlabel('X',fontsize=18) # legenda eixo x
plt.ylabel('Contagens por 0.30',fontsize=18) # legenda eixo y
plt.title('Histograma de variável gaussiana', fontsize=18) # título do histograma
bin_customizado = np.arange(-4.,4.,0.30) # criando bin de tamanho customizado

counts, bins, bars = plt.hist(y,bins=bin_customizado, edgecolor='black') # criando histograma
plt.show()
print(counts[12]) # pegando numero de contagens no bin 12
```

```python
lista_y = [] # criando lista para colocar os valores de y
bin_customizado = np.arange(-4.,4.,0.30) # repetindo procedimento anterior para criar distribuição aleatória
M = 0
mu=0
sigma=1
size=1000
#criando loop for para gerar multiplos valores de y e colocando na lista
while M <= 1000:
  y=np.random.normal(mu,sigma,size)

  counts, bins, bars = plt.hist(y,bins=bin_customizado,edgecolor="black")

  lista_y.append(counts[12])
  M+=1
print(lista_y)


```

```python
plt.xlabel('Y',fontsize=18) # legenda eixo x
plt.ylabel('Contagens de Y',fontsize=18) # legenda eixo y
plt.title('Histograma de Y', fontsize=18) # título do histograma
plt.hist(lista_y, bins=25, edgecolor='black') #criando histograma da lista
plt.show
```

```python
plt.xlabel('Y',fontsize=18) # legenda eixo x
plt.ylabel('$F_n(y)$',fontsize=18) # legenda eixo y
plt.title('Histograma normalizado de Y', fontsize=18) # título do histograma
grafico_freq_relativa = plt.hist(lista_y, bins=25, density=True, edgecolor='black') #normalizando histograma da lista
plt.show
```
